### Bronze Layer

```python
# ---- Medallion Architecture
# Load data from source ( Data Lake ) into dataFrame then write it into bronze layer
```

##### Read Data

```python
# Customers Dataset
customers_df = spark.read.format("csv") \
    .option('inferSchema',True) \
    .option('header',True)\
    .load('/Volumes/workspace/bronze_layer/source_data/customers_dataset.csv')

# Order_items Dataset
order_items_df = spark.read.format("csv") \
    .option('inferSchema',True) \
    .option('header',True)\
    .load('/Volumes/workspace/bronze_layer/source_data/order_items_dataset.csv')

# Orders Dataset
orders_df = spark.read.format("csv") \
    .option('inferSchema',True) \
    .option('header',True)\
    .load('/Volumes/workspace/bronze_layer/source_data/orders_dataset.csv')

# Products Dataset
products_df = spark.read.format("csv") \
    .option('inferSchema',True) \
    .option('header',True)\
    .load('/Volumes/workspace/bronze_layer/source_data/products_dataset.csv')

# Sellers Dataset
sellers_df = spark.read.format("csv") \
    .option('inferSchema',True) \
    .option('header',True)\
    .load('/Volumes/workspace/bronze_layer/source_data/sellers_dataset.csv')
```

##### Write data

```python

# Customers Dataset
customers_df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("workspace.bronze_layer.customers")

# Order_items Dataset
order_items_df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("workspace.bronze_layer.order_items")

# Orders Dataset
orders_df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("workspace.bronze_layer.orders")

# Products Dataset
products_df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("workspace.bronze_layer.products")

# Sellers Dataset
sellers_df.write.format("delta") \
  .mode("overwrite") \
  .saveAsTable("workspace.bronze_layer.sellers")

```
